# JupyterMap Demo

Interactive MapLibre GL maps from the CPRA Master Plan Data API.

**Requirements:** `uv sync --extra jupyter`

```python
from cpra.mp.data.visualizations import JupyterMap
```

## Salinity: annual, single year

```python
JupyterMap(
    variable="sal",
    grid="hydro_compartment_v001",
    time_unit="annual",
    calendar_year=2047,
    scenario_id=7,
    model_group_id=609,
)
```

## Elevation — annual, single year

Raster-grid elevation data filtered to a specific calendar year.

```python
JupyterMap(
    variable="elev", grid="morph_pixel_v001", time_unit="annual", calendar_year=2030, scenario_id=7, model_group_id=500
)
```

## Inundation — annual, future year

```python
JupyterMap(
    variable="inun", grid="morph_pixel_v001", time_unit="annual", calendar_year=2050, scenario_id=7, model_group_id=500
)
```

## Multiple filter values

Pass a list for any filter param — encoded as repeated query params.

```python
JupyterMap(
    variable="elev",
    grid="morph_pixel_v001",
    time_unit="annual",
    calendar_year=[2019, 2070],
    scenario_id=7,
    model_group_id=500,
    show_projects=True,
)
```

## `show()` — explicit display inside functions/loops

Use `show()` when the map is not the last expression in a cell.

```python
years = [2020, 2040, 2070]

for year in years:
    print(f"--- {year} ---")
    m = JupyterMap(
        variable="elev",
        grid="morph_pixel_v001",
        time_unit="annual",
        calendar_year=year,
        scenario_id=7,
        model_group_id=500,
    )
    m.show()
```

## Inspect the map.json URL

`JupyterMap.url` exposes the full endpoint for debugging or manual fetch.

```python
m = JupyterMap(variable="elev", grid="morph_pixel_v001", time_unit="annual", calendar_year=2030)
print(m.url)
```
